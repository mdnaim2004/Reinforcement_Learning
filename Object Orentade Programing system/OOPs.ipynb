## Object Orentade Programing system

#### OOPs is a programing paradigm based based on:
- Classes -- blueprint
- Objects -- instances
- Attributes -- variable inside a class
- Methodes -- function inside a class

#### Main OOP principles:
- **Encaptulation** : Keeping data and methos together in a class and restracting direc access to internal data.
- **Inheritance** : One class drives from another, gainig its properties and behaviour(code reuse)
- **Polymorphism** : Same methode name behaving differently depending on the object calling it.
- **Abstrution** : Showing only essential feature while hiding unnecessary implemention details.

```python
# make new module...
def find_fib(n):
    if n<= 2:
        return 1
    fib_x, fib_next = 1, 1

    i = 3
    while i<= n:
        i+=1
        fib_x, fib_next = fib_next, fib_x + fib_next
        return fib_next

for x in range(1, 11):
    print(find_fib(x))
```

### Mini Resing Project of the OOPs

```python
class CAR:
    def __init__(self, name, manufructure, color, year, cc):
        self.name = name
        self.manufructure = manufructure
        self.color = color
        self.year = year
        self.cc = cc

    def start(self):
        print(f"{self.name} engine started 🚗")

    def brakes(self):
        print(f"{self.name} is braking 🛑")

    def turn(self, direction):
        print(f"{self.name} turning {direction} ↩️")

    def change_gear(self, gear):
        print(f"{self.name} changed to gear {gear} ⚙️")
```

```python
car1 = CAR("BMW", "Germany", "Black", 2022, 2000)

car1.start()
car1.change_gear(2)
car1.turn("left")
car1.brakes()
```

```python
class Vehicle:
    """Base class for all vehilees"""
    def __init__(self, name, manuf,color):
        self.name = name
        self.manuf = manuf
        self.color = color
    
    def drive(self):
        print("Driving the : ", self.manuf, self.name)
    
    def turn(self,direction):
        print("Turning the : ", self.name, "to ",direction)
    
    def brake(self):
        print("The ",self.name, " is stoping...!")

if __name__ == "__main__":
    v1 = Vehicle("BWM", "bmw manufructure Ltd.", "Black")
    v2 = Vehicle("Marcedis Benj", "mercidis automoobile manufructure Ltd.", "Dark Nevy_blue")
    v3 = Vehicle("Toyota", "toyota autombile manufructure Ltd.", "Black")

    v1.drive()
    v2.drive()
    v3.drive()

    v1.turn("left")
    v2.turn("right")

    v1.brake()
    v2.brake()
    v3.brake()
```
