```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A' : [1, 2, np.nan, 3],
    'B' : [2, np.nan, np.nan, 4],
    'C' : [7, 8, 9, 4]
}

df = pd.DataFrame(data)
df
```

### Nan value delete

```python
df.dropna()
```

```python
df
```

```python
df.dropna(axis = 1)
```

```python
df
```

```python
df.dropna(axis = 1, inplace=True)
```

```python
df
```

```python
data = {
    'A' : [1, 2, np.nan, 3],
    'B' : [2, np.nan, np.nan, 4],
    'C' : [7, 8, 9, 4]
}

df = pd.DataFrame(data)
df
```

```python
df.dropna(axis = 0)
```

## Threshold use

```python
df.dropna(axis = 0, thresh=2)
```

```python
df
```

## filling the datset in fillna using

```python
df.fillna(value = 'missing')
```

```python
df.fillna(value = df['C'].mean())
```

```python
df['A'].fillna(value = df['A'].mean(), inplace=True)
```

```python
df
```
