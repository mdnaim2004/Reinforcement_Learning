```python
import numpy as np
import pandas as pd

```

```python
left = pd.DataFrame(
    {
        'key1' : ['k0', 'k1', 'k2', 'k3'],
        'A' : np.random.randint(1,10, size = 4),
        'B' : np.random.randint(1,10, size = 4)

    }
)

right = pd.DataFrame(
    {
        'key1' : ['k0', 'k2', 'k4', 'k5'],
        'C' : np.random.randint(1,10, size = 4),
        'D' : np.random.randint(1,10, size = 4)

    }
)
```

```python
left
```

```python
right
```

```python
# By default it set inner join
# pd.merge(left,right, on= 'key1')
#or
pd.merge(left, right, on= 'key1', how = 'inner')
```

```python
pd.merge(left, right, on='key1', how= 'outer')
```

```python
pd.merge(left, right, on= 'key1', how= 'left')
```

```python
pd.merge(left, right, on='key1', how= 'right')
```

```python
left = pd.DataFrame(
    {
        'A' : np.random.randint(1,10, size = 4),
        'B' : np.random.randint(1,10, size = 4)

    }, index= [0,2,3,4]
)
right = pd.DataFrame(
    {
        'A' : np.random.randint(1,10, size = 4),
        'B' : np.random.randint(1,10, size = 4)

    }, index= [0,1,2,3]
)
```
