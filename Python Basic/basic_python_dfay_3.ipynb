## *Leap Year*

```python
year = input("Enter your Year : ")
```

```python
year = int(year)
```

```python
if year % 4 != 0:
    print("NO! this is not leap year.")
else:
    if year % 100 == 0:
        if year % 400 == 0:
            print("YES, this is leap year.")
        else:
            print("NO! this is not leap year.")
    else:
        print("YES, this is leap year.")
```

### Another way

```python
if year % 400 == 0:
    print("YES, this is leap year.")
elif year % 100 == 0:
    print("YES, this is not leap year.")
elif year % 4 == 0:
    print("YES, this is leap year.")
else:
    print("YES, this is leap year.")
```

###  *Rectangle*

```python
import turtle
turtle.shape("turtle")

turtle.forward(100)
turtle.left(90)
turtle.forward(100)
turtle.left(90)
turtle.forward(100)
turtle.left(90)
turtle.forward(100)

turtle.exitonclick()
```
