```python
def func(name):
    print("hello,from func....")
    print(name)
    
```

```python
arr = [1,2,3,4,5,6]
```

```python
def func(array):
    new_arr = []

    for num in array:
        if num % 2 == 0:
            new_arr.append(num)
    
    return new_arr
```
