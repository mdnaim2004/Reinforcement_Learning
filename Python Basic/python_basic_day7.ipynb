# Zip and Enumerate
---

## Zip

---

zip() is useful to combine multiple lists or other iterable element-wise.

```python
name = ['naim', 'saim', 'rahim', 'karim']
marks = [91, 89, 87, 95]

for name, marks in zip(name, marks):
    print(f"{name} : {marks}")
```

```python
name = ['naim', 'saim', 'rahim', 'karim']
marks = [91, 89, 87, 95,78, 95, 90]

for name, marks in zip(name, marks):
    print(f"{name} : {marks}")
```

```python
name = ['naim', 'saim', 'rahim', 'karim','hafiz']
marks = [91, 89, 87]

for name, marks in zip(name, marks):
    print(f"{name} : {marks}")
```

```python
name = ['naim', 'saim', 'rahim', 'karim']
marks = [91, 89, 87, 95]

student_marks = dict(zip(name, marks))
print(student_marks)
```

```python
name = ['naim', 'saim', 'rahim', 'karim','hafiz']
marks = [91, 89, 87]

student_marks = dict(zip(name, marks))
print(student_marks)
```

```python
for idx, student in enumerate(name, start=1):
    print(f"{idx}. {student}")
```
