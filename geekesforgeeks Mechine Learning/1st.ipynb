## Rule of Function

**1.Non-default parameters must come before default parameters in the function definition.**

মানে, যে parameter-এর default value নেই, সেগুলো আগে লিখতে হবে।

```python
def greet(name, age=20):
    pass
```

**২. Positional arguments must come before keyword arguments when calling a function.**

Function call করার সময় আগে positional argument, তারপর keyword argument দিতে হবে

```python
def add(a, b, c):
    print(a, b, c)
```

```python
add(1, 2, c=3)
```

**2. If using keyword arguments, order doesnot metter.**

Keyword ব্যবহার করলে order আর গুরুত্বপূর্ণ না।

```python
def person(name, age, city):
    print(name, age, city)
```

```python
person(name="Naim", city="Bogura", age=22)
```

```python
person(age=22, city="Rajshahi", name="Naim")
```

**৪. Each parameter must have only one value.**

একটা parameter-কে দুইবার value দেওয়া যাবে না।

```python
def add(a,b):
    print(f"The sum of {a} and {b} is = {a + b}")
```

```python
add(2,4)
```
