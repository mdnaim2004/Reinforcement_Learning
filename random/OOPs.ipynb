```python
class car:
    name = ""
    color = ""

    def __init__(self, name, color):
        self.name = name
        self.color = color

    def start(self):
        print("Starting the engine")

my_car = car("Mercidse", "Dark Nevy_Blue")
print(my_car.name)
print(my_car.color)

my_car.start()
```

```python
class car:
    n = ""
    c = ""

    def __init__(self, n, c):
        self.n = n
        self.c = c
    
    def start(self):
        print("The car engin start")

my_car = car("BMW", "Black")

print(my_car.n)
print(my_car.c)
my_car.start()
```
